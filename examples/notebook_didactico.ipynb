## 1. Qué modela Bohr

El modelo de Bohr describe un **átomo hidrogenoide**: núcleo de carga $Z$ y **un solo electrón** en órbitas circulares $n = 1, 2, 3, \ldots$

$$E_n = -13.6\,\mathrm{eV}\cdot\frac{Z^2}{n^2}, \qquad r_n = n^2\cdot\frac{a_0}{Z}$$

con $a_0 \approx 0.529\,\text{Å}$. Una transición $n_i \to n_f$ intercambia un fotón de energía $|\Delta E|$, con frecuencia $\nu = |\Delta E|/h$ y longitud de onda $\lambda = hc/|\Delta E|$.

**Sí entra:** H ($Z=1$), He⁺ ($Z=2$), Li²⁺ ($Z=3$).  
**No entra bien:** átomos con varios electrones (He neutro, C, O, …). Tampoco hay relatividad ni estructura fina. Si el desarrollo se expande, hay que **volver a validarlo**.

## 2. Instalar y crear el átomo

La clase `BohrAtom` encapsula las constantes (Rydberg, $a_0$, Planck, …) y el número atómico $Z$.

````python
from bohr_h_andresdperezc import BohrAtom

hidrogeno = BohrAtom(Z=1)
helio = BohrAtom(Z=2)
print(hidrogeno)
print("Z =", hidrogeno.Z)
print("Rydberg (eV) =", hidrogeno.ENERGIA_RYDBERG_EV)
print("a0 (Å) =", hidrogeno.RADIO_BOHR_A)
````

## 3. Energía y radio

Valores de libro para el hidrógeno: $E_1 = -13.6\,\mathrm{eV}$ y $r_1 = a_0$.  
Unidades por defecto: **eV** y **Å**.

````python
print(f"E_1 = {hidrogeno.energia(1):.2f} eV")
print(f"r_1 = {hidrogeno.radio(1):.3f} Å")
print()
print("n    E (eV)      r (Å)")
for n, e, r in zip(range(1, 5), hidrogeno.energias(4), hidrogeno.radios(4)):
    print(f"{n}  {e:10.4f}  {r:8.4f}")
````

## 4. Átomos hidrogenoides: $E \propto Z^2$

He⁺ tiene un electrón, como el hidrógeno, pero $Z=2$. El nivel fundamental debe ser cuatro veces más profundo.

````python
print(f"E_1(H)   = {hidrogeno.energia(1):.2f} eV")
print(f"E_1(He⁺) = {helio.energia(1):.2f} eV")
print(f"4 × E_1(H) = {4 * hidrogeno.energia(1):.2f} eV")
print(f"r_1(He⁺) = {helio.radio(1):.3f} Å  (la mitad de a0)")
````

## 5. Transiciones: emisión y absorción

Si el electrón **baja** de nivel, el átomo **emite** un fotón. Si **sube**, **absorbe**.  
La línea de Balmer $n=3 \to 2$ del hidrógeno está cerca de **656 nm** (H-alpha).  
La longitud de onda se reporta en **nm** y la frecuencia en **Hz**.

````python
emision = hidrogeno.transicion(3, 2)
absorcion = hidrogeno.transicion(2, 3)

print("3 → 2")
print(f"  tipo          = {emision.tipo}")
print(f"  ΔE            = {emision.delta_e:.4f} eV")
print(f"  energía fotón = {emision.energia_foton:.4f} eV")
print(f"  frecuencia    = {emision.frecuencia:.4e} Hz")
print(f"  λ             = {emision.longitud_onda:.1f} nm")
print()
print("2 → 3")
print(f"  tipo = {absorcion.tipo}")
print(f"  λ    = {absorcion.longitud_onda:.1f} nm  (la misma línea)")
````

## 7. Validaciones

$Z$ y $n$ deben ser enteros $\geq 1$. Una “transición” al mismo nivel no está permitida.

````python
for llamada in (
    lambda: BohrAtom(Z=0),
    lambda: hidrogeno.energia(1.5),
    lambda: hidrogeno.transicion(2, 2),
):
    try:
        llamada()
    except ValueError as error:
        print(error)
````
